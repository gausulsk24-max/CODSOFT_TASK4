```python

import pandas as pd
import re
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report

# 1. Dataset automatically download karein
url = "https://raw.githubusercontent.com/justmarkham/pycon-2016-tutorial/master/data/sms.tsv"
df = pd.read_table(url, header=None, names=['label', 'message'])

# 2. Text clean karein
def clean_text(text):
    return re.sub(r'[^a-zA-Z0-9\s]', '', text.lower())

df['clean_message'] = df['message'].apply(clean_text)
df['label_num'] = df['label'].map({'ham': 0, 'spam': 1})

# 3. Train-Test Split (80% Train, 20% Test)
X_train, X_test, y_train, y_test = train_test_split(
    df['clean_message'], df['label_num'], test_size=0.2, random_state=42
)

# 4. TF-IDF & Multinomial Naive Bayes Model
tfidf = TfidfVectorizer(stop_words='english')
X_train_vec = tfidf.fit_transform(X_train)
X_test_vec = tfidf.transform(X_test)

model = MultinomialNB()
model.fit(X_train_vec, y_train)

# 5. Accuracy check
y_pred = model.predict(X_test_vec)
print(f"Model Accuracy: {accuracy_score(y_test, y_pred) * 100:.2f}%\n")
print(classification_report(y_test, y_pred, target_names=['Ham', 'Spam']))

# 6. Live Testing Function
def predict_sms(msg):
    cleaned = clean_text(msg)
    vec = tfidf.transform([cleaned])
    res = model.predict(vec)[0]
    return "SPAM MESSAGE ⚠️" if res == 1 else "LEGITIMATE (HAM) ✅"

print("-" * 40)
print("Test 1:", predict_sms("Congratulations! You won a $1000 cash prize. Click here now."))
print("Test 2:", predict_sms("Hey, are we still meeting for dinner tonight?"))
```
